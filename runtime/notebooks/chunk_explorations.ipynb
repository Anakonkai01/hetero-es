```python
string_1 = "seed=0, chunk=0"
string_2 = "seed=0, chunk=1"

hash1 = hashlib.sha256(string_1.encode("utf-8")).digest()
hash2 = hashlib.sha256(string_2.encode("utf-8")).digest()
seed1 = int.from_bytes(hash1, "little")
seed2 = int.from_bytes(hash2, "little")
```

```python
random1 = np.random.Generator(np.random.PCG64(seed1)).standard_normal(4)
random2 = np.random.Generator(np.random.PCG64(seed2)).standard_normal(4)
print(random1)
print(random2)
```
